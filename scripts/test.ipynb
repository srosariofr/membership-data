```python
# import pandas as pd
import numpy as np

# 1. Sample Data: Companies with two dimensional scores (0 to 10)
data = {
    'Company': ['Alpha', 'Beta', 'Gamma', 'Delta', 'Epsilon', 'Zeta'],
    'Vision': [8.5, 4.2, 6.5, 3.1, 9.0, 5.5],
    'Execution': [7.0, 8.5, 5.0, 3.0, 8.0, 6.0]
}
df = pd.DataFrame(data)

# 2. Define Thresholds (Using Medians)
x_threshold = df['Vision'].median()
y_threshold = df['Execution'].median()


# 3. Calculate Quadrants
def get_quadrant(row, x_thresh=x_threshold, y_thresh=y_threshold):
    if row['Vision'] >= x_thresh and row['Execution'] >= y_thresh:
        return "Leaders"
    elif row['Vision'] < x_thresh and row['Execution'] >= y_thresh:
        return 'Challengers'
    elif row['Vision'] >= x_thresh and row['Execution'] < y_thresh:
        return "Visionaries"
    else:
        return "Niche Players" 


# Apply the function to create a Quadrant column
df['Quadrant'] = df.apply(get_quadrant, axis=1)

print(df)
print(x_threshold, y_threshold)
```
